```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
adult_census = pd.read_csv('data/numerical_data.csv')
```

```python
adult_census.head()
```

```python
target_name = "class"
target = adult_census[target_name]
target.head()
```

```python
data = adult_census.drop(columns=[target_name,"workclass","fnlwgt","education","education-num","marital-status","occupation","relationship","race","sex","native-country"])
data.head()
```

```python
data.columns
```

```python
data.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier()
_ = model.fit(data, target)
```

```python
target_predicted = model.predict(data)
```

```python
target_predicted[:5]
```

```python
target[:5] == target_predicted[:5]
```
